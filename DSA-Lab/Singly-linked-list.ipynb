```python
### Node class for singly linked list
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

# Function to create a linked list from user input
def create_linked_list():
    head = None
    tail = None
    n = int(input("Enter the number of nodes: "))
    for i in range(n):
        data = int(input(f"Enter data for node {i+1}: "))
        new_node = Node(data)
        if head is None:
            head = new_node
            tail = head
        else:
            tail.next = new_node
            tail = new_node
    return head

# Function to insert a node at a given position
def insert_node(head, data, position):
    new_node = Node(data)
    if position == 0:  # Insert at head
        new_node.next = head
        return new_node
    current = head
    count = 0
    while current and count < position - 1:
        current = current.next
        count += 1
    if not current:
        print("Position out of range.")
        return head
    new_node.next = current.next
    current.next = new_node
    return head

# Function to delete a node at a given position
def delete_node(head, position):
    if not head:
        print("List is empty.")
        return head
    if position == 0:  # Delete head
        return head.next
    current = head
    count = 0
    while current.next and count < position - 1:
        current = current.next
        count += 1
    if not current.next:
        print("Position out of range.")
        return head
    current.next = current.next.next
    return head

# Function to traverse and display the linked list
def traverse_list(head):
    if not head:
        print("List is empty.")
        return
    current = head
    while current:
        print(current.data, end=" -> ")
        current = current.next
    print("None")

# Menu-driven main program
if __name__ == "__main__":
    head = None
    while True:
        print("\n--- Singly Linked List Operations ---")
        print("1. Create Linked List")
        print("2. Insert Node")
        print("3. Delete Node")
        print("4. Traverse List")
        print("5. Exit")

        choice = input("Enter your choice: ")

        if choice == '1':
            head = create_linked_list()
            print("Linked List created successfully.")
        elif choice == '2':
            data = int(input("Enter data to insert: "))
            position = int(input("Enter position to insert (0-based index): "))
            head = insert_node(head, data, position)
            print("Node inserted successfully.")
        elif choice == '3':
            position = int(input("Enter position to delete (0-based index): "))
            head = delete_node(head, position)
            print("Node deleted successfully.")
        elif choice == '4':
            print("Current Linked List:")
            traverse_list(head)
        elif choice == '5':
            print("Exiting program.")
            break
        else:
            print("Invalid choice! Please try again.")
```
