# K-Nearest Neighbors Classification

This notebook classifies wine samples into the three known cultivar classes using chemical measurements from `Machine-Learning-Lab/data/wine-clustering.csv`. The train/test split and classifier are reproducible, and the evaluation uses macro-averaged multi-class metrics.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

# Support execution from either the repository root or the notebook directory.
repository_data_path = Path("Machine-Learning-Lab/data/wine-clustering.csv")
notebook_data_path = Path("../data/wine-clustering.csv")
data_path = next(
    (path for path in (repository_data_path, notebook_data_path) if path.is_file()),
    repository_data_path,
)

try:
    data = pd.read_csv(data_path)
except pd.errors.EmptyDataError as error:
    raise ValueError(
        f"The input file is empty: {data_path}. "
        "Add the wine-clustering.csv data before running the notebook."
    ) from error

wine = load_wine()
if len(data) != len(wine.target):
    raise ValueError(
        f"Expected {len(wine.target)} wine samples, but found {len(data)} in {data_path}."
    )

X = data.drop(columns=["Class"], errors="ignore")
y = pd.Series(wine.target, name="Class")

print(f"Loaded {len(data)} samples with {X.shape[1]} chemical measurements from {data_path}.")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y,
)

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)

metrics = {
    "Accuracy": accuracy_score(y_test, y_pred),
    "Precision (macro)": precision_score(y_test, y_pred, average="macro", zero_division=0),
    "Recall (macro)": recall_score(y_test, y_pred, average="macro", zero_division=0),
    "F1 score (macro)": f1_score(y_test, y_pred, average="macro", zero_division=0),
}

for metric_name, score in metrics.items():
    print(f"{metric_name}: {score:.4f}")
```
