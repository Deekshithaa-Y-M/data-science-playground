# Practical 01: Data Preprocessing

This notebook performs data cleaning for the ML lab dataset:
- handle missing numerical values with mean imputation
- handle missing categorical values with mode imputation
- remove duplicate records
- present cleaned features, target, null summary, and duplicate count

## Algorithm
1. Import pandas for data handling.
2. Load CSV from the repository data folder.
3. Fill missing numerical values using column means.
4. Fill missing categorical values using column mode.
5. Remove duplicate rows.
6. Split the cleaned dataset into features (X) and target (y).
7. Display cleaned outputs and validation summaries.

```python
from pathlib import Path
import pandas as pd
```

```python
# Resolve dataset path so the notebook works from either repo root or notebooks folder.
candidate_paths = [
    Path("../data/sample_input.csv"),
    Path("Machine-Learning-Lab/data/sample_input.csv"),
]

data_path = next((path for path in candidate_paths if path.exists()), None)
if data_path is None:
    raise FileNotFoundError(
        "Could not find sample_input.csv. Run this notebook from the repository root or the notebooks folder."
    )

data = pd.read_csv(data_path)
print(f"Loaded data from: {data_path.resolve()}")
data
```

```python
data.fillna(data.mean(numeric_only=True), inplace=True)

for col in data.select_dtypes(include="object"):
    data[col] = data[col].fillna(data[col].mode()[0])

data.drop_duplicates(inplace=True)

X = data.iloc[:, :-1].copy()
y = data.iloc[:, -1].copy()
```

```python
print("Features:\
", X.head())
print("\nTarget:\
", y.head())
print("\nMissing Values:\
", data.isnull().sum())
print("\nDuplicate Records:", data.duplicated().sum())
print("\nPreprocessed Dataset:\
", data.head())
```
